## Inspección visual de un producto

## Ficha PEAS

Percepción (S): matriz de 5x5 píxeles donde 0 representa un píxel normal y 1 representa un píxel defectuoso.

Acciones (A): aprobar, enviar a revisión manual o rechazar.

Entorno (E): línea de producción de una empresa.

Medida de desempeño (P): cantidad de productos correctamente clasificados y reducción de productos defectuosos enviados al cliente.


Objetivo: detectar productos defectuosos mediante la cantidad de píxeles defectuosos.


## Reglas de decisión

- 0 defectos: aprobar.
- De 1 a 3 defectos: enviar a revisión manual.
- Más de 3 defectos: rechazar.
