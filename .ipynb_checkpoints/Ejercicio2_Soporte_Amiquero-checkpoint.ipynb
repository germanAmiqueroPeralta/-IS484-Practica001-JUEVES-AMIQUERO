## ESCALAMIENTO DE TICKETS

## Ficha PEAS

Percepción (S): tiempo de espera, nivel de urgencia y condición de cliente premium.

Acciones (A): escalar a nivel 1, escalar a nivel 2 o escalar a nivel 3.

Entorno (E): sistema de mesa de ayuda de una empresa.

Medida de desempeño (P): tiempo de respuesta y correcta asignación de los tickets.

Objetivo: asignar cada ticket al nivel de soporte adecuado.


## Reglas de decisión

- Urgencia alta: escalar directamente a nivel 3.
- Urgencia media con espera mayor a 30 minutos: escalar a nivel 3.
- Urgencia media con cliente premium: escalar a nivel 3.
- Urgencia media en otros casos: escalar a nivel 2.
- Urgencia baja con espera mayor a 60 minutos: escalar a nivel 2.
- Urgencia baja en otros casos: escalar a nivel 1.
