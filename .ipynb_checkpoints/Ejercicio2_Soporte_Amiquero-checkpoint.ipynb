## ESCALAMIENTO DE TICKETS

## Ficha PEAS

Percepción (S): tiempo de espera, nivel de urgencia y condición de cliente premium.

Acciones (A): escalar a nivel 1, escalar a nivel 2 o escalar a nivel 3.

Entorno (E): sistema de mesa de ayuda de una empresa.

Medida de desempeño (P): tiempo de respuesta y correcta asignación de los tickets.

Objetivo: asignar cada ticket al nivel de soporte adecuado.


## Reglas de decisión

- Urgencia alta: escalar directamente a nivel 3.
- Urgencia media con espera mayor a 30 minutos: escalar a nivel 3.
- Urgencia media con cliente premium: escalar a nivel 3.
- Urgencia media en otros casos: escalar a nivel 2.
- Urgencia baja con espera mayor a 60 minutos: escalar a nivel 2.
- Urgencia baja en otros casos: escalar a nivel 1.

## Implementación de codigo

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):

    nivel_urgencia = nivel_urgencia.lower()

    if nivel_urgencia == "alta":
        return "escalar a nivel 3", "urgencia alta reportada por el cliente"

    elif nivel_urgencia == "media" and (tiempo_espera_minutos > 30 or cliente_premium):
        return "escalar a nivel 3", "urgencia media con espera alta o cliente premium"

    elif nivel_urgencia == "media":
        return "escalar a nivel 2", "urgencia media con tiempo de espera moderado"

    elif nivel_urgencia == "baja" and tiempo_espera_minutos > 60:
        return "escalar a nivel 2", "urgencia baja pero tiempo de espera elevado"

    else:
        return "escalar a nivel 1", "urgencia baja y tiempo de espera aceptable"
```
